# Project Title

**Netflix Analytics : Exploring Content Growth and Patterns**

# **Objective(Problem statment)**

This project aims to analyze Netflix content data to identify meaningful patterns, trends, and insights using exploratory data analysis (EDA).

The objective is to understand how Netflix content has grown over time and to explore patterns in genres, ratings, and content distribution across different countries.

The project also focuses on cleaning and preprocessing the dataset, creating meaningful visualizations, and providing actionable insights based on the analysis.

# Outcome

The analysis helps in identifying key trends in Netflix content growth and patterns in audience preferences.

It provides a clear understanding of content distribution, popular genres, and rating patterns.

The insights derived from this project can support better content strategy and decision-making.

# **Note For Kind of Analysis to be done**

**Type of Analysis**

* Descriptive Analysis : Descriptive analysis is used to summarize the dataset and understand the overall distribution of Netflix content.
It helps in identifying the total number of movies and TV shows, distribution of genres, ratings, and content across different countries and years.

* Diagnostic analysis : Diagnostic analysis is used to understand the reasons behind observed patterns in the dataset.
For example, it helps to analyze why certain genres are more popular, why specific countries produce more content, and what factors influence content distribution.

* Predictive analysis : Predictive analysis is used to identify future trends in Netflix content growth based on past data.
By analyzing year-wise trends, it helps in estimating how content production may grow in the future, such as predicting trends for upcoming years like 2026.

* Prescriptive analysis : Prescriptive analysis is used to provide recommendations based on the insights obtained from the data.
It helps in suggesting what type of content Netflix should focus on, which genres are performing well, and how content strategy can be improved.

## Stage 2 – Data Cleaning and Pre-processing

```python
# Filling duration with Unknown because it is a categorical/text column
df['duration'] = df['duration'].fillna("Unknown")
```

```python
# Filling categorical columns with meaningful values
df['country'] = df['country'].fillna("Unknown")
df['genres'] = df['genres'].fillna("Not Specified")
```

```python
# Filling text columns to avoid data loss
df['cast'] = df['cast'].fillna("Not Available")
df['director'] = df['director'].fillna("Not Available")
```

```python
# Dropping description column as it is not required for analysis
df = df.drop(['description'], axis=1, errors='ignore')
```

```python
# Checking duplicate records
df.duplicated().sum()
# No duplicate records found in this dataset
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Checking outliers using boxplot
sns.boxplot(x=df['release_year'])
plt.title("Outlier Detection - Release Year")
plt.show()
```

A boxplot was used to detect outliers in the release year column. The visualization shows that there are no significant outliers present in the dataset no outlier treatment was required



```python
# Data Type Conversion
# Converting date added to datetime format for better time based analysis
df['date_added'] = pd.to_datetime(df['date_added'])
```

```python
# Creating new column year added for analysis
df['year_added'] = df['date_added'].dt.year
```

```python
# Creating new column 'year_added' for analysis
df['year_added'] = df['date_added'].dt.year
```

```python
# Filtering dataset to include recent data (2022 onwards)
df = df[df['release_year'] >= 2022]
```
